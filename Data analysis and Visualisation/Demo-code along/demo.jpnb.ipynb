```python
orders['category'] = (
    orders['category']
    .str.strip()
    .str.lower()
    .replace('home and garden', 'home & garden')
    .str.title()
)
orders['category'].value_counts()
```

```python
orders['express'] = orders['express'].str.lower().map({"yes":True, "y":True, "no":False})
print(orders['express'].isna().sum())
orders['express'].value_counts()
```

```python
orders['unit_price'].astype(float)
```

```python
price = orders['unit_price'].str.strip()
print(price.head(3).tolist())

price = price.str.replace(' kr', '')
print(price.head(3).tolist())

price = price.str.replace(' ', '')
print(price.head(3).tolist())

price = price.str.replace(',', '.')
print(price.head(3).tolist())


```

```python
orders['unit_price'] = price.astype(float)
print(orders['unit_price'].dtype)
orders['unit_price'].describe().round()
```

```python
customers_raw = pd.read_csv('data/customers.csv')
customers = customers_raw.copy()
customers.shape
```

```python
customers.head(8)
```

```python
customers.info()
```

```python
customers['age'].describe()
```

```python
print((customers['age'] == 999).sum())
print((customers['age'] == 0).sum())
```

```python
print(customers.duplicated().sum())
print(customers['customer_id'].duplicated().sum())
```

```python
customers['age'] = customers['age'].where(customers['age'].between(15,100))

print(customers['age'].isna().sum())
```

```python
customers['age'].describe()
```

```python
print(customers.dropna().shape)
print(customers.dropna(subset=['city']).shape)
print(customers.dropna(subset=['age']).shape)
```

```python
customers['city'] = customers['city'].fillna('Unknown')
customers['city'].value_counts()
```

```python
age_median = customers['age'].median()
age_filled = customers['age'].fillna(age_median)

print(age_median)
print(customers['age'].mean().round(2), age_filled.mean().round(2))
print(customers['age'].std().round(2), age_filled.std().round(2))
```
