```python
import pandas as pd
listings = pd.read_csv("Data/listings.csv")
```

```python
listings["property_type"].value_counts().head(5)
```

```python
shares=listings['property_type'].value_counts(normalize=True).round(3)
shares
```

```python
listings['neighbourhood_cleansed'].value_counts().tail(3)
```

```python

print(listings.groupby('room_type')['beds'].mean().round(2))
(listings.groupby('room_type')['beds'].agg(['count', 'mean']).round(2))
```

```python
reviews = (
    listings.groupby('neighbourhood_cleansed')['number_of_reviews']
    .aggregate(['count', 'mean', 'median'])
    .round(2)
    .sort_values("median")
           )

reviews
```
