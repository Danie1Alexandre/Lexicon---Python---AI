```python
missing_values = df.isna().sum()
missing_values_percentage = (df.isna().mean()*100).round(2)
```

```python
missing_summary = pd.DataFrame({
    'missing values': missing_values,
    'missing percentage': missing_values_percentage
})

display(missing_summary[missing_summary['missing values']> 0].sort_values(by = 'missing values', ascending= False) )
```

There are several columns with missing data, but I only need to care about the columns that affect my analysis. Worldwide Gross only has 0.2% missing data, which makes it a good pick. 

US DVD Sales is 82% missing. DVD sales are not reported, or the movie was not released on DVD . However, DVDs are not part of my analysis, instead, I will use gross numbers, so I can just ignore it. I will drop US DVD Sales since I'm not using it.


Title had one missing value, which I fixed earlier by removing it. Removing one entry from 3,201 movies will most likely not have a big impact.

Major genre has 8.5% missing (not reported) values; I could rename them to "Unknown" so they can be included in a genre chart.

Worldwide Gross only has 0.2% missing data. I will keep it as it is, since 0.2% won't have a big impact on the results.
