```python
Loans_clean = loans_raw.copy()

Loans_clean['genre'] = (
    Loans_clean['genre']
    .str.strip()
    .str.lower()
    .replace('sci fi', 'science fiction')
    .replace('sci-fi', 'science fiction')
    .replace("children's", 'children')
    .str.title()
)

Loans_clean['returned'] = (
    Loans_clean['returned']
    .str.strip()
    .str.lower()
    .map({'yes': True, 'y':True, 'no': False, 'n': False})
)

Loans_clean['late_fee'] = (
    Loans_clean['late_fee']
    .str.replace("kr", "")
    .str.replace(",", ".")
    .str.strip()
    .astype(float)   
)

Loans_clean['loan_date'] = pd.to_datetime(Loans_clean['loan_date'], errors= 'coerce')

Loans_clean = Loans_clean.drop_duplicates().reset_index(drop=True)

Loans_clean = Loans_clean.drop(columns="notes")

print(Loans_clean.shape)


```
