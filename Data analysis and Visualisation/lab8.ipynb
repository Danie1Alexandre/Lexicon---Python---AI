Claim 3: "The market has grown: only 131 listings started in 2016, but 610 in
2024.

```python
first_review_counts = listings['first_review'].dt.year.value_counts().sort_index()
print(first_review_counts)
```

```python
print("2016 count:", first_review_counts.get(2016, 0))
print("2024 count:", first_review_counts.get(2024, 0))
```

Claim 3: Hold, because every year since 2013 shows an increase in listing numbers; in 2026, they have a larger share of the market. Only in 2020 was there a decrease. This shows a trend of increased website activity . Yes , some listings from 2013 may have been taken down, but the overall trend of more listings remains.

Claim 4: "Most hosts in Stockholm are professional companies with many
listings."


```python
host =listings.groupby('host_id').size()
print("total hosts",len(host))

print(f"Single listing: {(host == 1).mean():.1%}")
print(f"two listing: {(host == 2).mean():.1%}")
print(f"More than 5 listing: {(host >= 5).mean():.1%}")
```

```python
print(
    listings.groupby(['host_id'])
    .size()
    .sort_values(ascending=False)
    .head(10))
```

Claim 4 doesn't hold because 90% of the hosts only have one listing. So it's very far-fetched to say most hosts are professional companies if the majority only have one place for rent.
